```python
import pandas as pd
import nflreadpy as nfl

# Load play-by-play data for seasons 2021 to 2025
seasons = list(range(2021, 2026))
pbp_multi_year = nfl.load_pbp(seasons)

# Convert to a pandas DataFrame
pbp_df = pbp_multi_year.to_pandas()

# Display basic information and the first 5 rows
print(f"DataFrame Shape: {pbp_df.shape}")
display(pbp_df.head())
```

```python
print(pbp_df.columns.to_list())
```

```python
post_play_features = [
    'ep',
    'yards_gained',
    'first_down',
    'touchdown',
    'pass_touchdown',
    'rush_touchdown',
    'interception',
    'fumble',
    'fumble_lost',
    'sack',
    'complete_pass',
    'passing_yards',
    'rushing_yards',
    'receiving_yards',
    'score_differential_post',
    'posteam_score_post',
    'defteam_score_post',
    'posteam_score',
    'defteam_score',
    'success',
    'series_success',
    'series_result',
    'field_goal_result',
    'extra_point_result',
    'two_point_conv_result',
    'penalty_yards'
]

# Identify which columns to drop that actually exist in the DataFrame
cols_to_drop = [col for col in post_play_features if col in pbp_df.columns]

# Drop the columns
pbp_df_cleaned = pbp_df.drop(columns=cols_to_drop)

print(f"Original shape: {pbp_df.shape}")
print(f"New shape after removing post-play features: {pbp_df_cleaned.shape}")
print(f"Dropped {len(cols_to_drop)} columns.")
```
